```python
## 1. What is Machine Learning?
## Machine Learning is a subset of Artificial Intelligence where computers learn patterns from data and use those patterns to make predictions or decisions.
## 2. What is AI??
#Artificial Intelligence is a field of computer science that enables machines to perform tasks that normally require human intelligence, such as learning, reasoning, recognizing images, and understanding language.
#Example: A voice assistant understanding your question.
## 3. What is DL??
#Deep Learning is a subset of Machine Learning that uses neural networks with multiple layers to learn complex patterns from large amounts of data.
#Example: Face recognition.
##Dataset
#A dataset is a collection of data used to train, test, or analyze a machine-learning model.
##Feature
#A feature is an input variable used by the model to make a prediction.
##Labels
#The label or target is the output that the model tries to predict.
```

```python
## 2. Types of Machine Learning
### Supervised Learning
#Learning from labeled data.
### Unsupervised Learning
#Finding patterns in unlabeled data.
### Reinforcement Learning
#Learning through actions, rewards and penalties.
```

```python
## import libraries
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
```

```python
iris = load_iris()
x=pd.DataFrame(iris.data, columns=iris.feature_names)
y=pd.Series(iris.target,name="target")
x.head()
```

```python
# check dataset size
print("Number of rows:",x.shape[0])
print("number of features:",x.shape[1])
```

```python
print("Features")
print(x.columns)
print("Null Values")
print(x.isnull().sum())
print(y.head())
print("\n Target Values")
print(y.unique())
```

```python
# train and test and split
X_train, X_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)
print("Training data",X_train.shape)
print("Testing data",X_test.shape)
```
